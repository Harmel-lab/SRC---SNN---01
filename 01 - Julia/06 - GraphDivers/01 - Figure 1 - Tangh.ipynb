```julia
##############################################################
## Graphique Article 01 - Tangh et linéarisaon par Mrceaux  ##
##############################################################

using Plots
using LaTeXStrings
```

```julia
# differential computation time
#------------------------------
const bbegin::Float32=-4.0
const bEnd::Float32=4.0
const bStep::Float32=0.005
const TableSize::Int32 = ((bEnd-bbegin)/bStep)+1
```

```julia
# Initialization of the computation vectors
#------------------------------------------
xVal=Vector{Float32}(undef,TableSize)
TanghVal=Vector{Float32}(undef,TableSize)
TanghFval=Vector{Float32}(undef,TableSize)
ErrorVal=Vector{Float32}(undef,TableSize)

xVal=zeros(TableSize)
TanghVal=zeros(TableSize)
TanghFval=zeros(TableSize)
ErrorVal=zeros(TableSize)

pointeur::Int32=1
# computation of the three equations h, zst and hs
# the injected current must be lower than 3
for i in bbegin:bStep:bEnd
 
    #-- TangH  --
    TanghVal[pointeur] = tanh(i)

    #-- TangH piecewise  --
    if i<=-4/3
        TanghFval[pointeur]=-1
    elseif i>-4/3 && i<4/3
        TanghFval[pointeur]=i*3/4
    elseif i>=4/3
        TanghFval[pointeur]=1
    end

    #-- error
    ErrorVal[pointeur] = TanghVal[pointeur] - TanghFval[pointeur]
    
    xVal[pointeur]=i
    pointeur+=1
end
```

```julia
# ── Graphiques gauche ──────────────────────────────────────────

# Signal 1 — avec rectangle + flèche vers le zoom
fig = plot(xVal,TanghVal,
           size   = (1200, 750),
           xlims=(-4,4),
           legendfontsize = 20,
           tickfontsize = 20,    # nombres sur les axes
           label = "Tanh",
           linecolor = :blue,
           linewidth=2,
           left_margin   = 6Plots.mm,
           right_margin  = 3Plots.mm,
           top_margin    = 3Plots.mm,
           bottom_margin = 3Plots.mm)

plot!(xVal,TanghFval,labeltextsize = 20, linewidth=2,label = "PWF", linecolor = :green)
plot!(xVal,ErrorVal, linewidth=2,label = "Error", linecolor = :red)

savefig("TanhFunction-01_New.pdf")
```

```julia
##############################################################
## Graphique Article 01 - Comparaison Src_Flo VS Src_FPGA   ##
##############################################################

using Plots
using LaTeXStrings

include("../MyLib/Fpgalib-02.jl")
```

```julia
# differential computation time
#------------------------------
const bEnd::Float32=35.0
const bStep::Float32=0.1
const TableSize::Int32 = bEnd/bStep
```

```julia
# use of NeSRC3Fpga with 5 arguments - R, rs, bias, Zmaw and Zmin (Fpgalib-02)
# ----------------------------------------------------------------------------
# Initialization of the computation vectors - Florent format - float
#-------------------------------------------------------------------
h=Vector{Float32}(undef,TableSize)
hs=Vector{Float32}(undef,TableSize)
zs=Vector{Float32}(undef,TableSize)
x=Vector{Float32}(undef,TableSize)

h=zeros(TableSize)
hs=zeros(TableSize)
zs=zeros(TableSize)
x=zeros(TableSize)

#condition initiale
ht::Float64=0.0
ht1::Float64=0.0
hst::Float64=0.0
hst1::Float64=0.0
zst::Float64=0.1
xt::Float64=-1

pointeur::Int32=1

# Initialization of the computation vectors - FPGA format - INT
#--------------------------------------------------------------
Fh=Vector{Int16}(undef,TableSize)
Fhs=Vector{Int16}(undef,TableSize)
Fzs=Vector{Int16}(undef,TableSize)
Fx=Vector{Int16}(undef,TableSize)

Fh=zeros(TableSize)
Fhs=zeros(TableSize)
Fzs=zeros(TableSize)
Fx=zeros(TableSize)

#condition initiale
Fht::Int16=0
Fht1::Int16=0
Fhst::Int16=0
Fhst1::Int16=0
Fzst::Int16=100
Fxt::Int16=-1000

# Initialization of the initial conditions and parameters
#--------------------------------------------------------
const r::Float32 = 2  # constant for the Update part - updated value
const rs::Float32 = -7
const bias::Float32 = -5
const zshyp::Float32 = 0.91 # 0.9345
const zsdep::Float32 = 0.0

const Fparam=Vector{Int16}(undef,5)
Fparam[1] = -3000
Fparam[2] = 903
Fparam[3] = 100

# computation of the three equations h, zst and hs
# the injected current must be lower than 3
for i in 1.0:bStep:bEnd
 
    if ((pointeur>50) && (pointeur<250))
        xt=1 
        Fxt=1000 
    else 
        xt=-1
        Fxt=-1000
    end

    #-- FLOAT Florent --
    ht1 = tanh(2*xt+r*ht+rs*hst+bias)
    h[pointeur] = ht1
    
    zst =zshyp-(zshyp-zsdep)*(1/(1+exp(-10*(ht-0.5))))
    zs[pointeur] = zst

    hst1 = zst*hst+(1-zst)*ht
    hs[pointeur] = hst1
    
    ht = ht1
    hst = hst1
    #-- INT VHDL --
    Fx[pointeur] = Fxt
    Fh[pointeur],Fhs[pointeur],Fzs[pointeur]=NeSRC3Fpga5!(Fxt,Fht,Fhst, Fparam)

    Fht=Fh[pointeur]
    Fhst=Fhs[pointeur]
 
    #-- Next point --
    x[pointeur] = xt
    pointeur+=1
end
```

```julia
xplot=x[1: 300]
hplot=h[1: 300]
Fhplot = Fh[1: 300]

# ── Zones de zoom ──────────────────────────────────────────────
z2 = (230, 250)
z3 = (230, 250)

mask2 = z2[1] .<= xplot .<= z2[2]
mask3 = z3[1] .<= xplot .<= z3[2]

# ── Graphiques gauche ──────────────────────────────────────────

# Signal 1 — avec rectangle + flèche vers le zoom
p1 = plot(x[1:300], xlims=(0,300), color=:steelblue, label=false, 
          title="Current", ylabel="Amplitude [a.u.]")

# Signal 2 — avec rectangle + flèche vers le zoom
p2 = plot(h[1:300], xlims=(0,300),color=:red, label=false, 
          title="h(Zhyp=0.910)", ylabel="Amplitude [a.u.]")

rect2 = Shape([z2[1], z2[2], z2[2], z2[1]],
              [-1.05, -1.05, 1.05, 1.05])

plot!(p2, rect2,
      fillalpha=0.15, fillcolor=:red,
      linecolor=:red, linestyle=:dash, label=false)

plot!(p2,
    [z2[2]+5, z2[2] + 45],
    [0, +0.3],
    arrow     = arrow(:closed, 0.04, 0.10),
    color     = :red,
    linewidth = 1.0,
    label     = false)

# Signal 3 — sans zoom
p3 = plot(Fh[1:300], xlims=(0,300),color=:green, label=false, 
          title="h(Zhyp=902)", ylabel="Amplitude [a.u.]", xlabel="Time (cycle)")

rect3 = Shape([z3[1], z3[2], z3[2], z3[1]],
              [-1050, -1050, 1050, 1050])

plot!(p3, rect3,
      fillalpha=0.15, fillcolor=:green,
      linecolor=:green, linestyle=:dash, label=false)

plot!(p3,
    [z3[2]+5, z3[2] + 45],
    [0, 0],
    arrow     = arrow(:closed, 0.04, 0.10),
    color     = :green,
    linewidth = 1.0,
    label     = false)

# ── Zooms droite ───────────────────────────────────────────────
zoom1 = plot([z2[1]:z2[2]], h[z2[1]:z2[2]],
    title="Zoom h(Zhyp=0.910)", xlabel="Time (cycle)", ylabel="Amplitude [a.u.]",
    color=:red, label=false ,
    xlims=z2, framestyle=:box,
    linewidth=2)

zoom2 = plot([z2[1]:z2[2]], Fh[z3[1]:z3[2]],
    title="Zoom h(Zhyp=902)", xlabel="Time (cycle)", ylabel="Amplitude [a.u.]",
    color=:green, label=false,
    xlims=z3, framestyle=:box,
    linewidth=2)

# ── Assemblage ─────────────────────────────────────────────────
l = @layout[
        [grid(3, 1){0.95h, 0.65w}   grid(2, 1){0.95h, 0.35w} ]
        [f{0.05h, 0.65w} g{0.05h, 0.35w}]
    ]

label_a = plot(
    framestyle = :none,
    grid = false,
    legend = false,
    ticks = nothing,
    xlims = (0, 1),
    ylims = (0, 1),
    annotations = [(0.5, 0.5, text("(a)", 18))]
)

label_b = plot(
    framestyle = :none,
    grid = false,
    legend = false,
    ticks = nothing,
    xlims = (0, 1),
    ylims = (0, 1),
    annotations = [(0.5, 0.5, text("(b)", 18))]
)

blank = plot(framestyle = :none,
             legend = false,
             grid = false,
             ticks = nothing)

fig = plot(p1, p2, p3,zoom1, zoom2, label_a, label_b,
           layout = l, 
           size   = (1200, 750),
           left_margin   = 3Plots.mm,
           right_margin  = 3Plots.mm,
           top_margin    = 3Plots.mm,
           bottom_margin = 3Plots.mm)

savefig("Figure03.pdf")
```
